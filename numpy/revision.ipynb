```python
import numpy as np

students = ["Chala", "Aanaa", "Mimi", "Jerry"]

scores = np.array([
    [80, 90, 70],
    [60, 75, 85],
    [95, 88, 92],
    [70, 65, 80]
])

subjects = ["Python", "Math", "Database"]
```

```python
average=np.mean(scores,axis=1)
for idx in range(len(students)):
    if average[idx]>=80:
        print(f"{students[idx]},{average[idx]:.2f}")
high=np.max(scores,axis=0)
for idx in range(len(subjects)):
    print(f"{subjects[idx]},{high[idx]}")
```
