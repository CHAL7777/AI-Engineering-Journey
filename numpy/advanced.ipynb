```python
import numpy as np

numbers = np.array([12, 5, 8, 20, 15, 3, 10, 8, 25, 7])

scores = np.array([
    [78, 85, 90],
    [55, 60, 72],
    [92, 88, 95],
    [40, 50, 45],
    [67, 73, 70],
    [88, 79, 84]
])

students = ["Chala", "Aanaa", "Mimi", "Jerry", "Sara", "Lensa"]
subjects = ["Math", "Physics", "Chemistry"]
```

```python
average=np.mean(scores,axis=1)
```
