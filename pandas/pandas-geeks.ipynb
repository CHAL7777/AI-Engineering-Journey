```python


data = {
    "Name": ["Abel", "Sara", "Daniel", "Hana", "Miki"],
    "Age": [20, 21, 19, 20, 22],
    "Math": [85, 72, 95, 60, 78],
    "Biology": [90, 88, 92, 75, 83]
}

df = pd.DataFrame(data)

# 1. Number of students
print("Number of students:", df.shape[0])

# 2. Average Math score
print("Average Math score:", df["Math"].mean())

# 3. Student with the highest Math score
best_student = df.loc[df["Math"].idxmax()]
print("\nTop Math student:")
print(best_student)

# 4. Students scoring at least 80 in Math
high_scores = df.loc[df["Math"] >= 80]
print("\nStudents scoring at least 80:")
print(high_scores)

# 5. Average score across both subjects for each student
df["Average"] = df[["Math", "Biology"]].mean(axis=1)

print("\nStudent averages:")
print(df[["Name", "Average"]])
```
