```python
import pandas as pd

data = {
    "Name": ["Chala", "Aanaa", "Mimi", "Jerry", "Sara"],
    "Python": [85, 90, 75, 95, 60],
    "NumPy": [80, 85, 90, 88, 70],
    "Pandas": [90, 95, 85, 92, 65]
}

df = pd.DataFrame(data)

print(df)
```

```python
new_df=df.sort_values(by=["Python","NumPy"],ascending=[False,True])
```

```python
df["Python_rank"]=df["Python"].rank(ascending=False,)
df["NumPy_rank"]=df["NumPy"].rank(ascending=False)
```
